# Eksperimen Data Mining dengan Naive Bayes
Pada eksperimen ini, kita akan menggunakan algoritma Naive Bayes untuk memprediksi apakah suatu kejadian menghasilkan korban luka (kolom `# Injured` > 0) berdasarkan dataset.

```
# Import library yang diperlukan
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
```

```
# Muat dataset ke dalam DataFrame
file_path = 'accidental_deaths_children.csv'
data = pd.read_csv(file_path)
# Tampilkan beberapa baris pertama
data.head()
```

## 1. Preprocessing Data
Kita perlu mempersiapkan data dengan mengubah fitur menjadi format numerik dan memilih fitur relevan.

```
# Buat kolom target 'Injured' (True jika ada korban luka)
data['Injured'] = data['# Injured'] > 0

# Pilih fitur untuk model
features = ['State', '# Killed']
target = 'Injured'

# One-hot encoding untuk kolom 'State'
data_encoded = pd.get_dummies(data[features])

# Bagi data menjadi train (70%) dan test (30%)
X_train, X_test, y_train, y_test = train_test_split(
    data_encoded, data[target], test_size=0.3, random_state=42
)
```

## 2. Training Model Naive Bayes
Melatih model Gaussian Naive Bayes dengan data latih.

```
# Inisialisasi dan pelatihan model
model = GaussianNB()
model.fit(X_train, y_train)
```

## 3. Evaluasi Model
Menampilkan confusion matrix sebagai diagram heatmap untuk visualisasi hasil.

```
# Lakukan prediksi pada data uji
y_pred = model.predict(X_test)

# Hitung confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Plot confusion matrix
fig, ax = plt.subplots()
im = ax.imshow(cm)

# Atur label sumbu
labels = ['Tidak Luka', 'Luka']  # kelas False dan True
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(labels)
ax.set_yticklabels(labels)
ax.set_xlabel('Prediksi')
ax.set_ylabel('Sebenarnya')

# Tambahkan nilai di tiap sel
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, cm[i, j], ha='center', va='center')

plt.title('Diagram Confusion Matrix')
plt.show()
```
